# 03. Machine Learning Benchmarking & Dual-Validation Analysis

Examines the **42.10 percentage point generalization gap** between:
- **Subject-Dependent (SD)**: Stratified K-Fold cross-validation (training and test data share participants).
- **Subject-Independent (SI)**: Group K-Fold cross-validation (held-out subject evaluation).

Reproduces the empirical summary table published in the study.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_res = pd.read_csv('../results/summary.csv')
print(df_res.to_string(index=False))
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df_res, x='Model', y='Average', hue='Validation', palette={'SD': '#3b82f6', 'SI': '#ef4444'})
plt.title('Subject-Dependent vs. Subject-Independent Accuracy Comparison')
plt.ylabel('Average Accuracy (%)')
plt.xlabel('Classification Model / Ensemble')
plt.ylim(30, 105)
plt.axhline(50, color='gray', linestyle=':', label='Chance Baseline (50%)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
