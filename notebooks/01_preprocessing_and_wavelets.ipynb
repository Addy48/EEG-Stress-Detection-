# 01. EEG Signal Preprocessing & Wavelet Denoising

This notebook demonstrates the preprocessing pipeline used in **'Inter-Subject Variability in EEG-Based Stress Detection'**:
1. Baseline drift removal using Savitzky-Golay filtering.
2. Daubechies-2 discrete wavelet decomposition (4 levels) and VisuShrink soft thresholding.
3. Comparative inspection of raw vs. preprocessed EEG epochs.


```python
import numpy as np
import matplotlib.pyplot as plt
from src.preprocessing import remove_baseline_drift, wavelet_denoise, preprocess_trial

# Generate representative synthetic EEG epoch (14 channels, 2 seconds @ 128 Hz)
fs = 128
n_samples = 256
t = np.linspace(0, 2, n_samples)

np.random.seed(42)
clean_signal = np.sin(2 * np.pi * 10 * t) + 0.5 * np.sin(2 * np.pi * 22 * t)
drift = 1.2 * np.sin(2 * np.pi * 0.2 * t) + 0.4 * t
noise = 0.3 * np.random.randn(n_samples)
raw_eeg = clean_signal + drift + noise

# Step 1: Baseline drift removal
drift_removed = remove_baseline_drift(raw_eeg, window_length=51, polyorder=3)

# Step 2: Wavelet denoising
denoised = wavelet_denoise(drift_removed, wavelet='db2', level=4)

print('Preprocessing completed successfully.')
```

```python
plt.figure(figsize=(12, 6))
plt.plot(t, raw_eeg, label='Raw EEG (with drift & noise)', alpha=0.6, color='gray')
plt.plot(t, drift_removed, label='Savitzky-Golay Filtered', color='orange', linestyle='--')
plt.plot(t, denoised, label='Wavelet Denoised (db2, VisuShrink)', color='navy', linewidth=2)
plt.title('EEG Signal Preprocessing Pipeline Verification')
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude (µV)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
